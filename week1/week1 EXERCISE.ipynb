```python
# imports
import os
import openai
import ollama
from dotenv import load_dotenv
```

```python
# constants
load_dotenv(override=True)
OPENAI_API_KEY = os.getenv('OPENAI_API_KEY')

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2'
```

```python
# set up environment
```

```python
# here is the question; type over this to ask something new

question = """
Please explain what this code does and why:
yield from {book.get("author") for book in books if book.get("author")}
"""
```

```python
# Get gpt-4o-mini to answer, with streaming
def ask_openai(question):
    client = openai.OpenAI(api_key=OPENAI_API_KEY)
    response = client.chat.completions.create(
        model=MODEL_GPT,
        messages=[{"role": "user", "content": question}]
    )
    return response.choices[0].message.content
```

```python
# Get Llama 3.2 to answer
def ask_ollama(question):
    response = ollama.chat(model=MODEL_LLAMA, messages=[{"role": "user", "content": question}])
    return response['message']['content']
```

```python
# Your question
question = """
Please explain what this code does and why:
yield from {book.get("author") for book in books if book.get("author")}
"""

# Get response from GPT-4o-mini
gpt_response = ask_openai(question)
print("🔹 GPT-4o-mini Response:\n", gpt_response)

# Get response from Llama 3.2
llama_response = ask_ollama(question)
print("\n🔹 Llama 3.2 Response:\n", llama_response)
```
